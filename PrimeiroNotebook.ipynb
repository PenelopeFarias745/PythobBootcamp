```python
# Estruturas de seleção (condicionais)
#if
#elif
#else
#switch / case
```

```python
j = int(input('Valor de j: '))
k = int(input('Valor de k: '))

if j > k:
  # indentação obrigatória
  print(f'{j} maior do que {k}')

if j < k:
  print(f'{j} é menor do que {k}')

if j == k:
  print(f'{j} é igual a {k}')

# != --> diferente de
```

Exemplo:

Insira o nome, sobrenome, idade e altura do usuário

Em seguida:
- Escreva uma mensagem de saudação.
- Verifique se o usuário tem mais ou menos que 18 anos.
- Verifique se o usuário tem mais ou menos que 1.65m.

```python
nome = input('Insira o nome: ')
sobrenome = input('Insira o sobrenome: ')
idade = int(input('Insira a idade: '))
altura = float(input('Insira a altura (em metros): '))

print('Olá!')

if idade > 18:
  print('O usuário é maior de idade.')
else:
  print('O usuário é menor de idade.')

if altura > 1.65:
  print('O usuário tem mais que 1.65m.')
else:
  print('O usuário tem menos que 1.65m.')
```
